## Create a Text File


```python
import json

input_json = fr"D:\VATEX_ne\english_captions\vatex_val_en.json"
output_txt = "vatex_val_en.txt"

with open(input_json, "r", encoding="utf-8") as f:
    data = json.load(f)

with open(output_txt, "w", encoding="utf-8") as f:
    for item in data:
        for cap in item["caption"]:
            f.write(cap.strip() + "\n")

print(".txt created")
```

## Convert Text File To Nepali JSON File comparing with English Captions JSON File

```python
import json

original_json = fr"D:\VATEX_ne\english_captions\vatex_val_en.json"
translated_txt = "vatex_val_ne.txt"
output_json = "vatex_val_ne.json"

with open(original_json, "r", encoding="utf-8") as f:
    data = json.load(f)

with open(translated_txt, "r", encoding="utf-8") as f:
    ne_caps = [line.strip() for line in f if line.strip()]

idx = 0

for item in data:
    num_caps = len(item["caption"])  
    item["caption"] = ne_caps[idx: idx + num_caps]
    idx += num_caps

assert idx == len(ne_caps), "Mismatch between captions and translations!"

with open(output_json, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print("captions_ne.json created successfully")
```

## Split the text file for transalation


```python
def split_text_file(input_file, max_lines=50000):
    with open(input_file, "r", encoding="utf-8") as f:
        lines = f.readlines()

    part = 1
    for i in range(0, len(lines), max_lines):
        output_file = f"captions_en_part{part}.txt"
        with open(output_file, "w", encoding="utf-8") as out:
            out.writelines(lines[i:i + max_lines])
        print(f"Created {output_file} with {len(lines[i:i + max_lines])} lines")
        part += 1


split_text_file("vatex_train_en.txt", max_lines=50000)
```

## Merge the Splitted Part After Translation

```python
translated_files = [
    "captions_en_part1-ne_NP.txt",
    "captions_en_part2-ne_NP.txt",
    "captions_en_part3-ne_NP.txt",
    "captions_en_part4-ne_NP.txt",
    "captions_en_part5-ne_NP.txt",
]

with open("vatex_train_ne.txt", "w", encoding="utf-8") as out:
    for f in translated_files:
        with open(f, "r", encoding="utf-8") as infile:
            out.writelines(infile.readlines())

print("Merged translated captions into captions_ne.txt")
```
