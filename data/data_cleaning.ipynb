## Checking whether the video ids of captions.json file exits in the videos folder or not


```python
import json
import os

# Paths
video_folder = fr"D:\VATEX_ne\videos_train_val"
train_json_path = fr"nepali_captions\vatex_train_ne.json"
val_json_path = fr"nepali_captions\vatex_val_ne.json"

def extract_video_ids(json_path):
    with open(json_path, 'r',encoding="utf-8") as f:
        data = json.load(f)
    
    video_ids = set()

    if isinstance(data, list):
        for item in data:
            if "video_id" in item:
                video_ids.add(item["video_id"])
    elif isinstance(data, dict):
        for vid in data.keys():
            video_ids.add(vid)

    return video_ids

# Extract unique video IDs
train_ids = extract_video_ids(train_json_path)
val_ids = extract_video_ids(val_json_path)

all_ids = train_ids.union(val_ids)

print(f"Total unique video IDs: {len(all_ids)}")

# Check existence
missing_videos = []

for vid in all_ids:
    video_path = os.path.join(video_folder, f"{vid}.mp4")
    
    if not os.path.exists(video_path):
        missing_videos.append(vid)

# Results
print(f"Missing videos: {len(missing_videos)}")

# Print missing IDs
for vid in missing_videos:
    print(vid)
```

## Reverse Check

```python
import json
import os


# Paths
video_folder = fr"D:\VATEX_ne\videos_train_val"
train_json_path = fr"nepali_captions\vatex_train_ne.json"
val_json_path = fr"nepali_captions\vatex_val_ne.json"


def extract_video_ids(json_path):
    with open(json_path, 'r',encoding="utf-8") as f:
        data = json.load(f)
    
    video_ids = set()

    
    if isinstance(data, list):
        for item in data:
            if "video_id" in item:
                video_ids.add(item["video_id"])
    elif isinstance(data, dict):
        
        if all(isinstance(v, list) for v in data.values()):
            video_ids = set(data.keys())
        
        elif "annotations" in data:
            for item in data["annotations"]:
                video_ids.add(item["video_id"])

    return video_ids


# Get all video IDs from JSON
train_ids = extract_video_ids(train_json_path)
val_ids = extract_video_ids(val_json_path)

all_caption_ids = train_ids.union(val_ids)

print(f"Total caption video IDs: {len(all_caption_ids)}")

# Get video IDs from folder
video_files = os.listdir(video_folder)

video_ids = set()
for file in video_files:
    if file.endswith(".mp4"):
        vid = os.path.splitext(file)[0]  
        video_ids.add(vid)

print(f"Total videos in folder: {len(video_ids)}")

# Reverse check
missing_in_json = []

for vid in video_ids:
    if vid not in all_caption_ids:
        missing_in_json.append(vid)

# Results
print(f"Videos without captions: {len(missing_in_json)}")

for vid in missing_in_json:
    print(vid)
```
