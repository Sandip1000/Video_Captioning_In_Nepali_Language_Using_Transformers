```python
import json
import matplotlib.pyplot as plt

# Load JSON file
with open(fr"d:\Model_for_DEMO\tf_qf_mbart_vatex\checkpoint-8000\trainer_state.json", "r") as f:
    data = json.load(f)

# Extract log history
log_history = data["log_history"]

# Lists for plotting
train_steps = []
train_loss = []

eval_steps = []
eval_loss = []

# Parse logs
for log in log_history:
    # Training loss
    if "loss" in log:
        train_steps.append(log["step"])
        train_loss.append(log["loss"])

    # Validation / evaluation loss
    if "eval_loss" in log:
        eval_steps.append(log["step"])
        eval_loss.append(log["eval_loss"])

# Plot
plt.figure(figsize=(10, 6))

plt.plot(train_steps, train_loss, label="Training Loss", marker='o',markersize = 3, color = 'orange')
plt.plot(eval_steps, eval_loss, label="Validation Loss", marker='s', markersize = 3, color = 'purple')

plt.xlabel("Steps")
plt.ylabel("Loss")
plt.title("Training and Validation Loss Curve")
plt.legend()

plt.show()
```
